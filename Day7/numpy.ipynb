```python
#create a 4x4 array and extract its first row and last column
#expected output:
#first row: [1 2 3 4]
#last column: [ 4  8 12 16]
import numpy as np

arr = np.arange(1,17).reshape(4,4)
first_row = arr[0]
print("first row:", first_row)

last_column = arr[:, -1]
print("last column:", last_column)
```

```python
# extract odd rows and even columns
#problem brakdown:
#rows: in programming, indexing starts from 0 odd-indexed rows would be 1 and 3
#columns: even-indexed columns would be 0 and 2
import numpy as np
#original array
samplearray = np.array([[3, 6, 9, 12],
                       [15, 18, 21, 24],
                       [27, 30, 33, 36], 
                       [39, 42, 45, 48]])

#using slicing to select odd rows and even columns
#the slice [1::2] selects every second element starting from index 1 (odd rows)
#the slice [0::2] selects every second element starting from index 0 (even columns)
newArray = samplearray[1::2, 0::2]
print("Original array:")
print(samplearray)
print("\nNew array with odd rows and even columns:")
print(newArray)
```

```python
# Stack Array horizontally
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
stacked_array = np.hstack((a, b))
print("Stacked array horizontally:", stacked_array)
```

```python
#stack array vertically
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
stacked_array = np.vstack((a, b))
print("Stacked array vertically:", stacked_array)
```
