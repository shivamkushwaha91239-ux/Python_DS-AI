```python
# python libararies
import pandas as pd
import matplotlib.pyplot as plt

pd.__version__
```

```python
#create a dataset
data ={
    "Name": ["Aman","Shivam","Vansh","riya","rahul","priya"],
    "Age":[21 ,20 ,12 ,22 ,23 ,19 ],
    "Marks":[85,95,67,76,89,56],
    "Study_Hours":[5,6,7,8,9,4]

}
#convert into dataframe
df = pd.DataFrame(data)
print("full dataset : \n",sep="")
print(df)
```

```python
#Head(print 5 rows)
print("\n first 5 rows using head()")
print(df.head())
```

```python
#Tail(print last
print("\n last 2 rows using tail()")
print(df.tail(2))
```

```python
# dataset information
print("\n Datset Information ")
print(df.info())
```

```python
# Statistical Description
print("\nStatistical description")
print(df.describe())
```

```python
#mean of marks
print("\nAverage marks")
print(df["Marks"].mean())
```

```python
#Max Marks
print("\nHighest Marks")
print(df["Marks"].max())
```

```python
#Minimum Marks
print("\nlowest Marks")
print(df["Marks"].min())
```

```python
# Filter Students whose Scored more than 80
#Max Marks
print("\nStudents Scoring more than 80")
print(df[df["Marks"]>80])
```

```python
#correlation between Study hours and marks
#Max Marks
print("\nCorrelation")
print(df[["Study_Hours","Marks"]].corr())
```

```python
#plot graph
plt.scatter(df["Study_Hours"],df["Marks"])
plt.title("Study_Hours  vs Marks")
plt.xlabel("Study_Hours")
plt.ylabel("Marks")
plt.show()
```
